```python
## if else statement 
age = int(input("Enter your age: "))

if age>=18:
    print("you are eligible")
else:
    print("not eligible")
```

```python
## elif statement

age = int(input("Enter your age: "))

if age<=12:
    print("you are child")
elif age<=18:
    print("you are teenager")
else:
    print("Your Adult")
```

```python
## Nested Conditional Statement
x=int(input("Enter any Number"))

if x>=0:
    if x%2==0:
        print("Positive Even")
    else:
        print("Positive Odd")
else:
    print("Negative Number")
```

```python
### While Loop 
### it will be excuted until the condition is true

count=0

while count<5:
    print(count)
    count=count+1
```

```python
count=0
while count%2==0:
    print(count)
    count=count+1
```

```python
### Loop Control Statement
###Break

for i in range(11):
    if i==5:
        break
    print(i)
```

```python
###Continue
for i in range(11):
    if i%2==0:
        continue
    print(i)
```

```python
###Pass

for i in range(5):
    if i==2:
        pass
    print(i)
```

```python
### Nested Loops 

for i in range(2):
    for j in range(3):
        print(f"i:{i} and j:{j}")
```

```python
#### Example sum of first n natural numbers using for and while loop
## While Loop

sum=0
count=1
n=10

while count<=n:
    sum=count+sum
    count=count+1

print(sum)
```

```python
### For Loop
sum=0
n=10
for i in range(11):
    sum=sum+i

print(sum)   
```
