```python
### List Comprehension
### using basic Syntax
square=[num**2 for num in range(10)]
print(square)
```

```python
[num**2 for num in range(10)]
```

```python
### with conditional logic
even_numbers=[num for num in range(10) if num%2==0]
print(even_numbers)
```

```python
### Nested List Comprehension
lst1=[1,2,3,4,5]
lst2=["a","b","c","d","e"]
pair=[(i,j) for i in lst1 for j in lst2]
print(pair)
```

```python
### Print Comprehension with Fuction Cells 
words=["Hello","Python","List","Comprehension"]
lenghts=[len(word) for word in words]
print(lenghts)
```

```python
### Mathematical Operations
set1={1,2,3,4,5}
set2={4,5,6}
union_set=set1.union(set2) ## both set in one set
print(union_set)
                     
intersection_set=set1.intersection(set2)  ### Only the common elements
print(intersection_set)

intersection_update=set1.intersection_update(set2)
print(intersection_update)

difference_set=set1.difference(set2)
print(difference_set)

symmetric_difference=set1.symmetric_difference(set2)
print(symmetric_difference)
```

```python
### Sets Method
set1.issubset(set2)
```

```python
set1.issuperset(set2)
```
