```python
### Single Parameter
def even_or_odd(num):
    if num%2==0:
        print("The number is even")
    else:
        print("The Number is odd")
```

```python
### Positional Arguments
def print_numbers(*args):
    for i in args:
        print(i)

```

```python
print_numbers(1,2,3,4,5,"Krish")
```

```python
def keyword_args(**kwargs):
    for  key,value in kwargs.items():
        print(f"{key}:{value}")
```

```python
keyword_args(name="Krish",age="22",height="5.10")
```

```python
###For both *args, **kwargs
def print_details(*args,**kwargs):
    for i in args:
        print(i)
    for  key,value in kwargs.items():
        print(f"{key}:{value}")
```

```python
print_details(11,2,4,name="krish",age="22")
```

```python
### Temp Conversion
def temp_conversion(temp,unit):
    if  unit=="C":
        return temp*9/5+32
    elif unit=="F":
        return (temp-32)*5/9
    else:
        return None
    
```

```python
temp_conversion(52,"C")
```

```python
### Password Strenght Checker
def pass_strength(password):
    if len(password)<8:
        return False
    if not any(char.isdigit() for char in password):
        return False
    if not any(char.islower() for char in password):
        return False
    if not any(char.isupper() for char in password):
        return False
    if not any(char in '!@#$%^&*()_+' for char in password):
        return False
    return True
```

```python
pass_strength("Deva@123")
```

```python
### Shopping Cart Cost
def Cart_cost(cart):
    total_cost=0
    for item in cart:
        total_cost+=item['Price']*item['Quantity']
    return total_cost

cart=[
        {'name':'Apple','Price':0.5,'Quantity':4},
        {'name':'Banana','Price':0.7,'Quantity':3},
        {'name':'Pineapple','Price':1.0,'Quantity':8}
    ]

total_costing=Cart_cost(cart)
print(total_costing)
```

```python
### Checking Palindrome
def Palindrome(s):
    s=s.lower().replace(" ","")   
    return s==s[::-1]

print(Palindrome("A man a plan a canal Panama"))
print(Palindrome("A where we go"))
```

```python
### Factorial Calculation

def factorial(n):
    if n==0:
        return 1
    else:
        return n*factorial(n-1)
factorial(3)
```

```python
def count_word_frequency(filepath):
    word_count = {}
    # Fixed: changed file_path to filepath to match the function argument
    with open(filepath, 'r') as file:
        for line in file:
            words = line.split()  
            for word in words:
                # Fixed: added () to lower()
                word = word.lower().strip('.,!?;:”\’')
                word_count[word] = word_count.get(word, 0) + 1

    return word_count

# Fixed: Wrapped the file name in quotes as a string
filepath = 'sample.txt' 
word_frequency = count_word_frequency(filepath)
print(word_frequency)
```
